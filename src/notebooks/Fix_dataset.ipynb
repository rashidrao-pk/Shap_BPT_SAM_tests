```python
from pathlib import Path

folder = Path('/Users/rashid/data/PhD/datacloud_data/repos/XAI/ShapBPT-SAM/shap_bpt_sam/examples/data/risultati_120/120_refined_filled')
dry_run = False  # keep True to preview; set False to actually rename

files = sorted(folder.glob('*_filled.npy'))
rename_pairs = []
skipped_already_refined = []
skipped_existing_target = []

for src in files:
    if src.name.endswith('_refined_filled.npy'):
        skipped_already_refined.append(src)
        continue

    dst = src.with_name(src.name.replace('_filled.npy', '_refined_filled.npy'))
    if dst.exists():
        skipped_existing_target.append((src, dst))
        continue

    rename_pairs.append((src, dst))

print(f'Folder: {folder}')
print(f'Matching *_filled.npy files: {len(files)}')
print(f'To rename: {len(rename_pairs)}')
print(f'Already refined: {len(skipped_already_refined)}')
print(f'Skipped because target exists: {len(skipped_existing_target)}')

print('\nPreview:')
for src, dst in rename_pairs[:10]:
    print(f'{src.name} -> {dst.name}')
if len(rename_pairs) > 10:
    print(f'... {len(rename_pairs) - 10} more')

if dry_run:
    print('\nDry run only. Set dry_run = False and rerun this cell to rename files.')
else:
    for src, dst in rename_pairs:
        src.rename(dst)
    print(f'Completed renaming {len(rename_pairs)} files.')
```

## Bring all `*_filled.npy` files into `120_refined_filled`


```python
from pathlib import Path
import shutil

src_dir = Path('/Users/rashid/data/PhD/datacloud_data/repos/XAI/ShapBPT-SAM/shap_bpt_sam/examples/data/risultati_120/120_fill')
dst_dir = Path('/Users/rashid/data/PhD/datacloud_data/repos/XAI/ShapBPT-SAM/shap_bpt_sam/examples/data/risultati_120/120_refined_filled')

dry_run = False      # keep True to preview; set False to actually copy/move
move_files = False  # False = copy files; True = move files and remove from src_dir

files = sorted(src_dir.glob('*_filled.npy'))
dst_dir.mkdir(parents=True, exist_ok=True)

transfer_pairs = []
skipped_existing = []

for src in files:
    dst = dst_dir / src.name
    if dst.exists():
        skipped_existing.append((src, dst))
        continue
    transfer_pairs.append((src, dst))

print(f'Source: {src_dir}')
print(f'Destination: {dst_dir}')
print(f'Matching *_filled.npy files: {len(files)}')
print(f'To transfer: {len(transfer_pairs)}')
print(f'Skipped because target exists: {len(skipped_existing)}')

print('\nPreview:')
for src, dst in transfer_pairs[:10]:
    action = 'MOVE' if move_files else 'COPY'
    print(f'{action}: {src.name} -> {dst}')
if len(transfer_pairs) > 10:
    print(f'... {len(transfer_pairs) - 10} more')

if dry_run:
    print('\nDry run only. Set dry_run = False and rerun this cell to transfer files.')
else:
    for src, dst in transfer_pairs:
        if move_files:
            shutil.move(str(src), str(dst))
        else:
            shutil.copy2(src, dst)
    action = 'Moved' if move_files else 'Copied'
    print(f'{action} {len(transfer_pairs)} files.')
```
